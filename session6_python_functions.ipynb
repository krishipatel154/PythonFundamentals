# Python Functions

## What are functions?
functions are bolck of code which are created to do particular task and we can reuse it.

1> abstraction

2> decomposition

## Benifites

- code modularity
- code readibility
- code reusability

```python
# create function
def is_even(num):
  """
  This function check whether the number is even or odd!!
  here, num = any valid integer
  """
  if type(num) == int:
    if num%2==0:
      print("it's even")
    else:
      print("it's odd")

  else:
    print("only numbers are allowed!")

is_even("k")
```

```python
print(print.__doc__)
```

```python
# default argument
def is_even(num = 0):
  print(num)

is_even()

# positional argument
# keyword argument
is_even(num=20)
```

```python
# *args - (tuple)
# allow us to pass a variable number of non-keyword arguments to a functions.
def multiply(*args):
  product = 1
  for i in args:
    product = product*i
  return product

multiply(1,2,3,4,5)
```

```python
# **kwargs - (dictionary)
# **kwargs allows us to pass any number of keyword arguments
def display(**kwargs):
  for key, value in kwargs.items():
    print(key, value)

display(india='delhi', shrilanka='colombo')
```

## point to remember while using *args and **kwargs

- order of arguments matter(normal-> *args -> **kwargs)
- the word args and kwargs are only convention, you can use any name of your choice.

## Variable Scope

```python
# we can use gloable variables inside the functions but can not make changes in that.
def g(y):
    print(x)
    print(x+1)

x = 1
g(x)
print(x)
```

```python
def g(y):
    x += 1

x = 1
g(x)
print(x)
```

## Lambda Function

A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

## Difference between lambda function and normal function

- no name
- lambda has no return value
- not reusable
- labda is written in one line

```python
# sum of two numbers
a = lambda a, b: a+b
a(2,3)
```

```python
# check if string has 'a'
a = lambda s: 'a' in s
a('krishi')
```

```python
# odd or even
a = lambda a: 'even' if a%2==0 else 'odd'
a(5)
```

## Higher Order Function

A function which return a function or take a function as input.

```python
def square(x):
  return x**2

def transform(f, L):
  output = []
  for i in L:
    output.append(f(i))

  print(output)

L = [1,2,3,4,5]
transform(square, L)
```

```python
# use of the lambda function
def transform(f, L):
  output = []
  for i in L:
    output.append(f(i))

  print(output)

L = [1,2,3,4,5]
transform(lambda x: x**2, L)
```

## Map

```python
list(map(lambda x: x**2, [1,2,3,4,5]))
```

```python
# odd or even
list(map(lambda x: 'even' if x%2==0 else 'odd', [1,2,3,4,5]))
```

```python
stu = [
    {
        'name': 'krishi',
        'age': 20,
        'gender': 'female'
    },
    {
        'name': 'subhu',
        'age': 23,
        'gender': 'female'
    },
    {
        'name': 'rohan',
        'age': 20,
        'gender': 'male'
    }
]

list(map(lambda stu: stu['gender'], stu))
```

## Filter

```python
l = [1,2,3,4,5]

list(filter(lambda x: x>3, l))
```

```python
# fetch fruites starting with 'a'
fruites = ['apple', 'banana', 'cherry']
list(filter(lambda f: f[0] == 'a', fruites))
```
