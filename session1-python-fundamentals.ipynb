# Datatypes

```python
# Datatypes - [integer, string, float, boolean, complex number]
print(1000000000)
print("krishi")
print(10.2)
print(True)
print(5+6j)

# list - mutable
print([1,2,3,4,5])

# tuple - immutable
print((1,2,3,4,5))

# sets - only unique values
print({1,2,3,4,5})

# dictonary
print({"fname":"krishi", "lname":"patel"})

# None
k = None
print(k)
```

# Variables

```python
# variables
name = "krishi"
age = 20
print(name, age)
a = b = c = 10
print(a,b,c)
x, y, z = 10, 20, 30
print(x,y,z)


# static vs dynamic typing
# in python we do not need to declare the datatype of the variable. it auto detect the type of variable which is called dynamic typing.
demo = 20
print(type(demo))

# static vs dynamic binding
# in python we do not have any type casting functions, we can dynamically chage the type of the variable which is called dynamic binding
a = 20
print(a)
a = "name"
print(a)
```

# Type Conversion

```python
# Type conversion:- 
# convert one datatype in another is called type conversion. there are two types of type conversion in python: 
# 1. Implicit - interpreter do automatically 
# 2. Explicite - user perform type conversion
a = "888"
a = int(a)
print(type(a))
```

# Literals

```python
# Literals - raw value which we store in the variable
a = 0b010101 # Binary literal
b = 100 # Decimal literal
c = 0o310 # Octal literal
d = 0x12c # Hexadecimal literal

x = 20.5  # Float literal
y = 10+20j  # Complex literal

print(a, b, c, d)
print(x, y)
print(1.5e3)  # 1.5 * 10^3
```
