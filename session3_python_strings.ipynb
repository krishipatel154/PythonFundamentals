# Strings


```python
# string - strings are the sequence of the unicode characters.
# create string
a = 'krishi'
b = "krishi"
c = '''krishi'''
d = """krishi"""

print(a,b,c,d)

# access string - indexing
str1 = "hello! how are you?"
print(str1[3])
print(str1[-3])
# slicing
print(str1[0:6])
print(str1[0:2:3])   # step-size = 3

# reverse the string
print(str1[::-1])
```

```python
# edit the string - python strings are immutable means we can not change it.
# delete the string
s = "krishi patel"
del s
# print(s)
```

### Operations on string

1. Arithmetic Operations
2. Relational Operations
3. Logical Operations
4. Loops on string
5. Membership Operations

```python
# arithmetic operations
print('hello' + ' ' + 'world')
print('hello '*5)

# relational operations
print('krishi' == 'patel')
print('krishi' > 'patel')  # compare based on ascii values
print('patel' > 'krishi')
print('Krishi' > 'krishi')
```

### String Functions

```python
# len
print(len('krishi'))

# max
print(max('krishi'))

# min
print(min('krishi'))

# sorted - sort the chars based on ascii values
print(sorted('krishi patel'))
```

```python
# only string function
s = 'krishi patel'

print(s.capitalize())
print(s.title())
print(s.upper())
print(s.lower())
print('HellO WoRLD'.swapcase())
```

```python
s = "My name is krishi patel."

print(s.count('i'))

# if word not find then give -1
print(s.find('is'))
# if word not find then give error
print(s.index('M'))

print(s.endswith('.'))
print(s.startswith('My'))

# example
pdf_name = "my-pdf.docs"
if pdf_name.endswith('.pdf'):
  print("Yes, it's a pdf!!")

else:
  print("Not a pdf!!")
```

### String Format

```python
name = "krishi"
age = 20

print("Hello, my name is {} and i am {} years old".format(name, age))
print(f"hello, my name is {name} and i am {age} years old.")
```

```python
# split
s = "hello, how are you?"
print(s.split())
print(s.split('o'))

# join
print(" ".join(['hello,', 'how', 'are', 'you?']))
print("o".join(['hell', ', h', 'w are y', 'u?']))

# replace
print(s.replace('hello', 'heyy'))

# strip - remove extra space
s1 = "   krishi        "
s1.strip()
```

## Programs

```python
# find the length of the string without using len() function
s = str(input("Enter the string: "))
counter =0

for i in s:
  counter += 1

print("length of str: ",counter)
```

```python
# extract username from the given email - krishi@gmail.com - krishi
email = str(input("Enter email: "))
email_spilt = email.split('@')
print(email_spilt)

username = email_spilt[0]
print(username)
```

```python
# count the frequency of the particular character in a provided string
s = str(input("Enter string: "))
char = str(input("Enter char: "))
print(f"Frequency od {char} is: ", s.count(char))
```

```python
# write a program which can remove a particular character in provided string
s = str(input("Enter string: "))
char = str(input("Enter char: "))
s.replace(char,'')
```

```python
# string is palindrome or not
s = str(input("Enter string: "))
rev_str = s[::-1]

if s == rev_str:
  print("Palindrome")
else:
  print("not a palindrome")
```

```python
# write a program to count number of words in string without using split
s = str(input("Enter string: "))
counter = 0

for i in s:
  if i == " ":
    counter += 1
print(counter)
print("Total words: ",counter+1)
```

```python
# write a program to print all the words of a string without using split
s = str(input("Enter string: "))
temp = ""
Words = []

for i in s:
  if i != " ":
    temp += i
  else:
    Words.append(temp)
    temp = ""

Words.append(temp)
print(Words)
```

```python
# write a program to convert the string in title case without using title()
s = str(input("Enter string: "))
temp = ""
Words = []
List = []

for i in s:
  if i != " ":
    temp += i
  else:
    Words.append(temp)
    temp = ""

Words.append(temp)
print(Words)

for i in Words:
  List.append(i[0].upper() + i[1:].lower())

print(" ".join(List))
```
