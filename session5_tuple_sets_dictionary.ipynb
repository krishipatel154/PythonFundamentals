# Tuple

A tuple in python is similar to list. the difference between two is that we cannot change the elements of the tuples once it is assigned. which means tuples are immutable.

## Characteristics

- orderd
- immutable
- allows duplicates

```python
# create tuple
t = ()
t2 = (2,)
t3 = (1,2,3,4,2,5,4)
t4 = (1, 2.5, True, "krishi", [1,2,3], (2,3))
print(t, t2, t3)
print(t4)

# 2D tuple
t5 = ((1,2,3),(4,5,6))
print(t5)

# 3D tuple
t6 = (((1,2,3),(4,5,6)),((7,8,9)))
print(type(t6))
print(t6)
```

## Accessing items of tuple

- indexing
- slicing

```python
# indexing
t = (1,2,3,4,5,(6,7),[8,9])
print(t[0])
print(t[-1])
print(t[-1][0])
```

```python
# slicing
t = (1,2,3,4,5,6)
print(t[1:3])
print(t[::-1])
print(t[-3:-1])
print(t[0:-1])
```

```python
# editing and adding - tuples are immutable so we can not add or edit items in tuples
t = (1,2,3,4,5,6)
# t[0] = 100

a = (1,2,3)
a = a + (4,5)
print(a)
```

```python
# deleting - we can only delete the whole tuple, but we can not delete single item of the tuple
t = [1,2,3,4,5]
del t
# print(t)
```

## Tuple Operations

```python
# Arithmetic
t1 = (1,2,3,4,5)
t2 = (3,4,5,6,7)

print(t1+t2)
print(t1*5)

# membership
print(1 in t1)
print(1 not in t2)

# loops
for i in t1:
  print(i)
```

## Tuple Functions

```python
# max/min/len/sum/sorted
t1 = (1,2,3,4,5)
t2 = (3,4,5,6,7)

print(max(t1))
print(min(t1))
print(len(t1))
print(sum(t1))
print(sorted(t1, reverse=True))
```

```python
# count
t = (1,2,3,4,5,3,5)
print(t.count(5))
```

```python
# index
t = (1,2,3,5,4,3,2)
print(t.index(2))
```

```python
# tuple unpacking
a,b,c = (1,2,3)
print(a,b,c)

# a,b = (1,2,3)
# print(a,b)

a,b, *others = (1,2,3,4,5)
print(a,b,others)
```

```python
# tuple comperihension
t = (1,2,3,4,5)
l = [i for i in t]
print(l)
```

## Difference between list and tuple

- mutability: tuples are immutable and lists are mutable.
- speed: tuples are faster than list.
- memory: tuple consume less memory than list.

# Sets

A set is an unordered collection of items. Every set element is unique(no duplicates) and must be immutable(can not be changed). However, a set itself is mutable. We can add or remove items from it.

sets can also be used to perform mathematical set operations such as union, intersection, symmentric difference etc.

## Characteristics
- unordered
- immutable
- no duplicates
- can't contain mutable datatupes

```python
# empty set
s = set()
print(s)

# 1d set
s1 = {1,2,3}
print(s1)

# 2d set - 2d sets are not allowed

# hetro
s3 = {1,2.5,True,"krishi"}
print(s3)

# type conversion
s4 = set([1,2,3])
print(s4)

# set can't have mutable items
# s5 = {1,2,3,[4,5]}
# print(s5)
```

## Adding items

```python
s = {1,2,3,4}

# add
s.add(5)
print(s)

# update
s.update([6,7,8])
print(s)
```

## Delete items

```python
s = {1,2,3,4,5}

# discard
s.discard(5)
print(s)

# remove
s.remove(4)
print(s)

# pop - it delete item randamly
s.pop()
print(s)

# clear - clear the set
s.clear()
print(s)

# del
del(s)

# what is the difference between discard and remove?
# both delete the given item but the difference is when we pass any item which is not available set the remove give error and discard just ignore it.
```

## Set Operations

```python
s1 = {1,2,3,4,5}
s2 = {3,4,5,6,7}

# union(|)
print(s1 | s2)

# intersection(&)
print(s1&s2)

# difference (-)
print(s1-s2)
print(s2-s1)

# symmentric difference (^)
print(s1^s2)

# membership operators
print(1 in s1)
print(2 not in s2)

# loops
for i in s1:
  print(i)
```

## Set Functions

```python
# len/sum/min/max/sorted
s = {1,2,3,4,5,6,7,8}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s, reverse=True))
```

```python
# union
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
print(s1.union(s2))

# update -
s1.update(s2)
print(s1)
print(s2)
```

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# intersection
s1.intersection(s2)
print(s1)

# intersection_update
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference / difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)
print(s1)

# intersection_update
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symentric difference / symentric difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)
print(s1)

# intersection_update
s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# copy
s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
```

```python
# frozenset - it is immutable
fs = frozenset([1,2,3,4,5])
print(fs)

fs2 = frozenset([1,2,3,frozenset([4,5,6])])
print(fs2)
```

# Dictionary

Dictionary is a collection of key-value pair.

for ex: dict = {'name': 'krishi', 'age': 20}

## Characterstics

- Mutable
- indexing has no meaning
- key can't be repeted
- keys can't be mutable items

## Create Dictonary

```python
# empty dict
d = {}
print(d)

# 1D dict
d1 = {'name': 'krishi', 'age': 20}
print(d1)

# 2D dict
d2 = {'name':'krishi', 'class' :12, 'subjects' :{'math': 20, 'gujarati':10}}
print(d2)

# using dict()
d3 = dict([('name','krishi'), ('age', 20)])
print(d3)

# use immutable values as key
d4 = {(1,2,3):'demo', 'name':'krishi'}
print(d4)
```

## Access items

```python
d = {'name':'krishi', 'age':20}
print(d['name'])
print(d.get('age'))
```

## Add items

```python
d = {'name':'krishi', 'age':20}
d['gender'] = 'female'
print(d)
```

## Remove items

```python
d = {'name':'krishi', 'age':20, 'gender': 'female', 'weight': 40}
# pop
d.pop('age')
print(d)

# popitem - delete last key-value pair
d.popitem()
print(d)

# clear
d.clear()
print(d)

# del
# del(d['name'])
del(d)
```

## Dictionary Operations

- membership
- loops

```python
# membership
d = {'name':'krishi', 'age':20}
print('name' in d)
```

```python
# loops
d = {'name':'krishi', 'age':20}
for i in d:
  print(i, d[i])
```

## Dictionary Functions

```python
d = {'name':'krishi', 'age':20}

# len
print(len(d))

# sorted
print(sorted(d))

# min/max
```

```python
d = {'name':'krishi', 'age':20}
# items
print(d.items())
# keys
print(d.keys())
# values
print(d.values())
```
