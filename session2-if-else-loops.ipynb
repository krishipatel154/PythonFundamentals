# If-Else Statement

```python
# If-Else statement is used when user waht to perform any task based on the condition. if the condition is true, execute task otherwise ignore that code.
email = str(input("Enter the email: "))
password = str(input("Enter the password: "))

if email == "krishi@gmail.com" and password == "123456":
    print("Welcome back!!")

elif email == "krishi@gmail.com" and password != "123456":
    password = str(input("Enter password again: "))
    if email == "krishi@gmail.com" and password == "123456":
        print("Welcome back!!")

    else:
        print("opps, wrong credantials again!")

else:
    print("Wrong credantials!!")
```

```python
# find the maximum of three given numbers
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))
num3 = int(input("Enter num3: "))

if (num1 > num2) and (num1 > num3):
    print("Num1 is greater!!")

elif (num2 > num1) and (num2 > num3):
    print("Num2 is greater!!")

elif num1==num2==num3:
    print("All are equal!!")

else:
    print("Num3 is greater!!")
```

```python
# calculator
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2"))
ope = str(input("enter the operator: "))

if ope == '+':
    print("Sum is: ", num1+num2)

elif ope == '-':
    print("Sub is: ", num1-num2)

elif ope == '*':
    print("Multiplication is: ", num1*num2)

elif ope == '/':
    print("Devision is: ", num1+num2)

else:
    print("Enter valid values!!")
```

# Loops in python

```python
# while loop
num = int(input("Enter the number: "))
i = 1
while i <= 10:
    print(num * i)
    i += 1
```

```python
# while loop with else
x = 1
while x<3:
    print(x)
    x += 1

else:
    print("limit crossed")
```

```python
# guess the number game
import random
num = int(input("Enter the number: "))
original_num = random.randint(1,100)
counter = 1

while num != original_num:
    if num < original_num:
        print("guess grater number!!")

    else:
        print("Guess the small number!!")

    num = int(input("Enter the number: "))
    counter += 1

else:
    print("Congratulations!!")
    print("Total attempts ", counter)
```

# Nested Loops

```python
classes = [1,2,3]
names = ["k", "j", "r", "s", "a"]

for cls in classes:
    for name in names:
        print(f"student {name} from class {cls}")
```

```python
# pattern program1
# 1
# 1 2
# 1 2 3
num = int(input("Enter the number: "))

for i in range(1, num+1):
    for j in range(1, i+1):
        print(j, end=" ")
    print(end="\n")
```

```python
num = int(input("Enter the number: "))

for i in range(1, num+1):
    for j in range(1, i+1):
        print(j, end=" ")
    for k in range(i-1, 0, -1):
        print(k, end=" ")
    print(end="\n")

for i in range(num, 0, -1):
    for j in range(1, i+1):
        print(j, end=" ")
    for k in range(i-1, 0, -1):
        print(k, end=" ")
    print(end="\n")
```
